# HW2.1: Downloading the Hourly Wikipedia Event Data

Downloads the hourly `.parquet` Wikipedia event files from the public `s3://dsan6000-wikipedia` bucket into a local `data/` folder.

```python
import os
import boto3
```

```python
SOURCE_BUCKET = "dsan6000-wikipedia"
SOURCE_PREFIX = "hourly_parquet/"
LOCAL_DATA_DIR = "data"

os.makedirs(LOCAL_DATA_DIR, exist_ok=True)
```

```python
s3_client = boto3.client("s3")

response = s3_client.list_objects_v2(Bucket=SOURCE_BUCKET, Prefix=SOURCE_PREFIX)
object_keys = [obj["Key"] for obj in response["Contents"] if obj["Key"].endswith(".parquet")]

print(f"Found {len(object_keys)} parquet files in s3://{SOURCE_BUCKET}/{SOURCE_PREFIX}")
object_keys
```

```python
for key in object_keys:
    filename = os.path.basename(key)
    local_path = os.path.join(LOCAL_DATA_DIR, filename)
    s3_client.download_file(SOURCE_BUCKET, key, local_path)
    print(f"Downloaded {key} -> {local_path}")
```

```python
downloaded_files = sorted(os.listdir(LOCAL_DATA_DIR))
print(f"{len(downloaded_files)} files now in {LOCAL_DATA_DIR}/")
downloaded_files
```
